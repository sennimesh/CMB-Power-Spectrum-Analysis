```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Multipole moments
ell = np.array([2, 10, 100, 500, 1000])

# Approximate angular scale in degrees
theta = 180 / ell

# Display the results
results = pd.DataFrame({
    "Multipole (ell)": ell,
    "Angular Scale (degrees)": theta
})

print(results)
```

```python

plt.figure(figsize=(8, 5))

plt.plot(ell, theta, marker="o")

plt.xscale("log")
plt.yscale("log")

plt.xlabel("Multipole Moment (ell)")
plt.ylabel("Angular Scale (degrees)")
plt.title("Angular Scale vs Multipole Moment")

plt.grid(True)
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

# Multipole range
ell = np.arange(2, 2001)

# Illustrative example spectrum (not real Planck data)
Cl = 1 / (ell * (ell + 1))

# Convert to D_l
Dl = ell * (ell + 1) * Cl / (2 * np.pi)

plt.figure(figsize=(10, 5))
plt.plot(ell, Dl)

plt.xlabel("Multipole Moment (ell)")
plt.ylabel("D_l (arbitrary units)")
plt.title("Illustrative Angular Power Spectrum")

plt.grid(True)
plt.show()
```
